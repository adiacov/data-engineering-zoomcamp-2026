```python
from pathlib import Path
import os

import logging

# logger = logging.getLogger(__file__)
logger = logging.getLogger("tmp")


def get_root_path() -> Path:
    """Returns the project rood Path object form a current working directory"""

    logger.info("Deriving root path from current working directory")
    root_path = None
    root_target = "data-engineering-zoomcamp-2026"
    cur_path = Path(os.getcwd())
    for parent in cur_path.parents:
        if parent.name == root_target:
            root_path = Path(parent)
            pass

    result_msg = (
        f"Successfully found root path from current working directory: {root_target}"
        if root_path
        else f"Could not find project root path from current working directory: {root_target}"
    )

    logger.info(result_msg)
    return root_path


get_root_path()
```
